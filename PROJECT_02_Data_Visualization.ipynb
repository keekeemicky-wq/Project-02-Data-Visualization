```python
# PROJECT 02: DATA VISUALIZATION WITH MATPLOTLIB & SEABORN

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load dataset
sns.set_theme(style="whitegrid", palette="deep")
df = sns.load_dataset("tips")

print("First 5 rows:")
print(df.head())

print("\nDataset information:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())

# 2. Bar Chart: Total Bill by Day
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="day", y="total_bill", hue="day",
            estimator=sum, palette="viridis", legend=False)
plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill (USD)")
plt.tight_layout()
plt.show()

# 3. Scatter Plot: Total Bill vs Tip
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="total_bill", y="tip",
                hue="smoker", palette="Set1", s=70)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill (USD)")
plt.ylabel("Tip (USD)")
plt.tight_layout()
plt.show()

# 4. Histogram: Bill Distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="total_bill", bins=20,
             kde=True, color="teal")
plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill (USD)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 5. Box Plot: Bill by Day
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="day", y="total_bill",
            hue="day", palette="Set2", legend=False)
plt.title("Total Bill Distribution by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill (USD)")
plt.tight_layout()
plt.show()

# 6. Correlation Heatmap
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm",
            fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

# 7. Average Tip by Party Size
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="size", y="tip",
            hue="size", estimator="mean",
            palette="mako", legend=False)
plt.title("Average Tip by Party Size")
plt.xlabel("Party Size")
plt.ylabel("Average Tip (USD)")
plt.tight_layout()
plt.show()

# 8. Correlation Analysis
print("\nCorrelation Matrix:")
print(correlation.round(2))

print("\nCorrelation between Total Bill and Tip:")
print(round(df["total_bill"].corr(df["tip"]), 3))

# 9. Key Findings
print("\nPROJECT CONCLUSION")
print("-------------------")
print("1. Bar charts compare total bill amounts across days.")
print("2. Scatter plots show the relationship between bills and tips.")
print("3. Histograms show the distribution of bill amounts.")
print("4. Box plots compare bill spread across days.")
print("5. Heatmaps reveal correlations between numerical variables.")
print("6. Custom palettes improve chart readability.")

print("\nNote: Correlation does not imply causation.")
```
